```python
from transformers import pipeline

pipe = pipeline("text-to-speech", model="suno/bark-small")
```

```python
fr_text = "♪ In the jungle, the mighty jungle, the ladybug was seen. ♪ "
output = pipe(fr_text, generate_kwargs={"max_length": None})
Audio(output["audio"].squeeze(), rate=output["sampling_rate"])
```

# Genrate Music:

```python
music_pipe = pipeline("text-to-audio", model="facebook/musicgen-small")
```

```python
text = "Pop music, something similar to the wee end song or M. Jackson"
forward_params = {"max_new_tokens": 512}

output = music_pipe(text, forward_params=forward_params)
```

```python
from IPython.display import Audio

output["audio"]
Audio(output["audio"], rate=output["sampling_rate"])
```
